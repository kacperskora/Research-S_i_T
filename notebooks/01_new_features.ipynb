# Adding new potentially (as cointegrated or as X) helpful time series

## Loading libraries

```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import datetime
```

## Loading datasets

```python
data = pd.read_csv('../data/raw/ready_merged_data_CO2.csv', parse_dates = ['Date'])
stoxx_600_data = pd.read_csv('../data/raw/stoxx.csv', parse_dates = ['Data'], decimal = ',')
data.info()
stoxx_600_data.info()
df = data.copy()
```

```python
stoxx_data = stoxx_600_data.copy()
stoxx_data = stoxx_data[['Data', 'Ostatnio']]
stoxx_data
```

```python
stoxx_data = stoxx_data.rename(columns = {
    'Data' : 'Date', 
    'Ostatnio' : 'stoxx_value'
})
stoxx_data
```

```python
stoxx_data['Date'] = pd.to_datetime(stoxx_data['Date'], format = '%d.%m.%Y')
stoxx_data.info()
stoxx_data
```

```python
matching_dates = df['Date'].isin(stoxx_data['Date']).sum()
print(matching_dates)
```

```python
full_df = pd.merge(df, stoxx_data, how = 'left', on = 'Date')
full_df.info()
```

```python
full_df.isna().sum()
```

```python
full_df['stoxx_value'].fillna(method = 'ffill', inplace = True)
full_df.info()
```

```python
full_df.to_csv('../data/raw/merged_df_new.csv')
```
